### 6.3. Sentiment Share by Time of Day

We group posting hours into four periods: Morning (5-12), Afternoon (12-17), Evening (17-21) and Night (21-5).

```python
def get_time_of_day(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df_main['Time_of_Day'] = df_main['Hour'].apply(get_time_of_day)
time_order = ['Morning', 'Afternoon', 'Evening', 'Night']

time_sentiment = pd.crosstab(df_main['Time_of_Day'], df_main['Sentiment_Group'], normalize='index') * 100
time_sentiment = time_sentiment.loc[time_order]
print(time_sentiment.round(1))
```

```python
time_sentiment.plot(kind='bar', color=['#ff7f0e', '#2ca02c', '#1f77b4'])

plt.title('Sentiment Share by Time of Day')
plt.xlabel('Time of Day')
plt.ylabel('Percentage of Posts (%)')
plt.xticks(rotation=0)
plt.legend(title='Sentiment Group')

plt.show()
```

**Takeaway:** Positivity rises through the day, from 55% in the morning to 65% in the evening. Night is the exception: it has the lowest positive share (49%) and the highest negative share (35%). However, night has the fewest posts (86), so this pattern should be treated as a tendency rather than a firm conclusion.

### 6.4. Sentiment and Engagement by Season

We group months into seasons (Winter: Dec-Feb, Spring: Mar-May, Summer: Jun-Aug, Fall: Sep-Nov) and compare both the sentiment mix and the average engagement.

```python
# Map months to seasons
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Fall'

df_main['Season'] = df_main['Month'].apply(get_season)
season_order = ['Winter', 'Spring', 'Summer', 'Fall']

# Sentiment share by season
season_sentiment = pd.crosstab(df_main['Season'], df_main['Sentiment_Group'], normalize='index') * 100
season_sentiment = season_sentiment.loc[season_order]
print(season_sentiment.round(1))
```

```python
season_sentiment.plot(kind='bar', color=['#ff7f0e', '#2ca02c', '#1f77b4'])

plt.title('Sentiment Share by Season')
plt.xlabel('Season')
plt.ylabel('Percentage of Posts (%)')
plt.xticks(rotation=0)
plt.legend(title='Sentiment Group')

plt.show()
```

```python
# Average engagement by season
print(df_main.groupby('Season')['Total_Engagement'].mean().loc[season_order].round(1))
```

**Takeaway:** Season shows the clearest temporal pattern. **Summer is the most positive season** (68% positive, only 19% negative) and also receives the highest engagement (about 70 interactions per post). **Fall is the least positive** (49% positive, 31% negative). Winter and spring fall in between.

### 9. Predictive Modeling

**Predict how many likes + retweets a post gets from its sentiment, time of day and platform, and compare the model with a simple baseline.**

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

```python
import warnings
warnings.filterwarnings("ignore", category=RuntimeWarning)
```

### 9.1. Prepare the Data

```python
model_df = df_main.copy()
model_df['Timestamp'] = pd.to_datetime(model_df['Timestamp'])
model_df['Time_of_Day'] = model_df['Timestamp'].dt.hour.apply(get_time_of_day)
```

```python
# Mark each post as "high engagement"
HIGH = model_df['Total_Engagement'].quantile(0.75)
model_df['high_engagement'] = (model_df['Total_Engagement'] >= HIGH).astype(int)

print(f"High engagement = {HIGH:.0f} or more likes + retweets", "\n")

print(model_df['high_engagement'].value_counts(), "\n") #If post got 75 or more likes + retweets = 1, otherwise 0.
```

**Above shows there are 240 posts that got 75 or more likes + retweets**

```python
# Turn text columns (sentiment, time of day, platform) into 0/1 columns the prediction model can use.
# A post with all zeros is the baseline: Negative, Afternoon, Facebook.

X_all = pd.get_dummies(model_df[['Sentiment_Group', 'Time_of_Day', 'Platform']], drop_first=True, dtype=int)
FEATURES = list(X_all.columns)

model_df = pd.concat([model_df.drop(columns=FEATURES, errors='ignore'), X_all], axis=1)
```

### 9.2. Train/Test Split (80% training / 20% testing)

```python
train, test = train_test_split(model_df, test_size=0.2, random_state=42,
                               stratify=model_df['high_engagement'])


print("Training posts:", len(train), "  Test posts:", len(test), "\n")
print(f"Share of high engagement  train: {train['high_engagement'].mean():.2%}   test: {test['high_engagement'].mean():.2%}\n")
```

### 9.3. Train and Evaluate the Model

```python
X_train, y_train = train[FEATURES], train['Total_Engagement']
X_test, y_test = test[FEATURES], test['Total_Engagement']

baseline_pred = np.full(len(test), y_train.mean())
line = LinearRegression().fit(X_train, y_train)

print("=== Regression: how many likes + retweets? ===")
print(f"Baseline MAE (always guess the average): {mean_absolute_error(y_test, baseline_pred):.1f}","\n")
print(f"Linear regression  train MAE: {mean_absolute_error(y_train, line.predict(X_train)):.1f}"
      f"   test MAE: {mean_absolute_error(y_test, line.predict(X_test)):.1f}")
```

### 9.4. What Drives Engagement?

```python
print("\nLearned weights (extra likes + retweets vs. a Negative, Afternoon, Facebook post):")
print(pd.Series(line.coef_, index=FEATURES).sort_values(ascending=False).round(1))
print("bias:", round(line.intercept_, 1))
```

### 9.5. Model vs. Baseline

```python
BLUE, ORANGE = "#2a78d6", "#eb6834"
rng = np.random.default_rng(0)

train_hat = line.predict(X_train)
test_hat = line.predict(X_test)

fig, ax = plt.subplots(figsize=(9, 5.5))

ax.scatter(train_hat + rng.uniform(-0.8, 0.8, len(train_hat)), y_train, s=18, alpha=0.35, color=BLUE, label='Training posts')
ax.scatter(test_hat + rng.uniform(-0.8, 0.8, len(test_hat)), y_test, s=30, alpha=0.8, color=ORANGE, marker='^', label='Test posts')

xs = np.linspace(train_hat.min() - 2, train_hat.max() + 2, 100)
ax.plot(xs, xs, color='black', linewidth=2, label='Learned model (its prediction)')
ax.axhline(y_train.mean(), color='gray', linestyle='--', linewidth=2,
           label=f'Baseline (always guess the average = {y_train.mean():.0f})')

ax.set(title='Learned model vs. baseline (always guess the average)',
       xlabel='Model prediction for the post (from tone, time of day and platform)',
       ylabel='Actual likes + retweets')
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

**Takeaway:** The model learned a real pattern (the line goes up), but the dots are spread widely, so it can’t predict each post exactly.

### 9.6. From Prediction to Strategy

```python
w = pd.Series(line.coef_, index=FEATURES)
start = line.intercept_

steps = [('Negative post\nafternoon, Facebook', start, 'start'),
         ('Make it\npositive', w['Sentiment_Group_Positive'], 'step'),
         ('Post it\nat night', w['Time_of_Day_Night'], 'step'),
         ('Post on\nInstagram', w['Platform_Instagram'], 'step')]
end = start + sum(v for _, v, kind in steps if kind == 'step')
steps.append(('Positive post\nat night, Instagram', end, 'end'))

fig, ax = plt.subplots(figsize=(11, 5.5))
level = 0
for i, (label, value, kind) in enumerate(steps):
    if kind == 'step':
        bottom, height, color = level, value, ('#2a78d6' if value >= 0 else '#eb6834')
        level += value
        text = f'{value:+.0f}'
        y_text = bottom + height
    else:
        bottom, height, color = 0, value, '#555555'
        level = value
        text = f'{value:.0f}'
        y_text = value
    ax.bar(i, height, bottom=bottom, color=color, width=0.6)
    ax.text(i, y_text + 1.5, text, ha='center', fontsize=14, fontweight='bold')
    if i < len(steps) - 1:
        ax.plot([i + 0.3, i + 0.7], [level, level], color='gray', linestyle=':', linewidth=1)

ax.set_xticks(range(len(steps)))
ax.set_xticklabels([s[0] for s in steps], fontsize=11)
ax.set_ylabel('Predicted likes + retweets')
ax.set_ylim(0, end * 1.2)
ax.set_title(f'Three simple choices take a post from about {start:.0f} to about {end:.0f} likes + retweets',
             fontsize=13)
for s in ['top', 'right']:
    ax.spines[s].set_visible(False)
plt.tight_layout()
plt.show()
```

**Takeaway:** Tone matters most, then timing; platform matters least. Making a post positive adds about 16 likes + retweets, posting at night adds about 11, and posting on Instagram adds about 5. Together, these three choices raise a post from about 49 to about 81.

Sentiment, timing and platform shift a post’s average engagement, but they cannot reliably predict a single post.
